 #### Overview
This supplementary material contains the complete Python codebase used to conduct the computational audit reported in the paper. The analysis is implemented across three Jupyter notebooks, one for each research question. The notebooks must be run in order as RQ3 depends on output files produced by the LPBA and MLM notebooks.

#### Model
###### ClinicalBERT
Both Revised_LPBA.ipynb and PA.ipynb use ClinicalBERT — a BERT-based masked language model pretrained on MIMIC-III clinical notes. The model downloads automatically on first run:

#### Model identifier: emilyalsentzer/Bio_ClinicalBERT
License: Apache 2.0
Source: https://huggingface.co/emilyalsentzer/Bio_ClinicalBERT
Paper: Alsentzer et al. (2019). Publicly available clinical BERT embeddings. arXiv:1904.03323.

An internet connection is required on first run (~440MB download). RQ3.ipynb does not use ClinicalBERT.

#### Running the Analysis
IMPORTANT — Run notebooks in this order:
1. Revised_LPBA.ipynb → 2. PA.ipynb → 3. RQ3.ipynb
RQ3 loads output files from Steps 1 and 2. Running out of order will produce NameError.

#### Step 1 — LPBA Analysis (Revised_LPBA.ipynb)
What it does:

Loads and filters MIMIC-III clinical notes by demographic group
Consolidates ethnicity labels into White, Black, Hispanic, Asian
Loads ClinicalBERT (emilyalsentzer/Bio_ClinicalBERT)
Extracts real clinical sentence templates (up to 20 per word, minimum 5 required)
Filters to patient-describing sentences only
Inserts demographic descriptors across eight race-gender combinations
Computes LPBA scores: log P_M(w|s,D_i) - log P_M(w|s,D_0)
Runs paired t-tests and intersectional analysis vs White Male reference
Generates heatmaps, forest plots, and diverging bar charts

#### Output files:

LPBA_revised_results.csv — full results
LPBA_significant_results.csv — race-level significant findings
LPBA_intersectional_results.csv — complete intersectional results
LPBA_intersectional_significant.csv — intersectional significant findings

#### Estimated runtime: 2–4 hours (GPU); 8–12 hours (CPU)

Step 2 — MLM Agency Attribution Analysis (PA.ipynb)
What it does:

Loads same MIMIC-III dataset with same demographic filtering
Defines agency attribution target words in three subcategories:

Active resistance: declined, refusing
Active cooperation: requested, agreed
Passive cooperation: responded, presented


Loads ClinicalBERT and extracts agency attribution sentence templates
Removes clinical deterioration sentences from declined
Inserts demographic descriptors across eight race-gender combinations
Computes MLM probabilities: P_M(w|s,D_i)
Runs paired t-tests vs White Male reference
Generates diverging bar chart, forest plot, heatmap

#### Output files:

MLM_full_results.csv — full results
MLM_significant_results.csv — significant findings
MLM_intersectional_results.csv — complete intersectional results

Estimated runtime: 2–4 hours (GPU); 8–12 hours (CPU)

#### Step 3 — Corpus Frequency and RQ3 Analysis (RQ3.ipynb)
What it does:

Loads MIMIC-III text directly — no ClinicalBERT needed
Counts target word frequency per demographic group
Normalizes to per 10,000 tokens
Loads LPBA and MLM significant results from Steps 1 and 2
Compares model probability direction against corpus frequency direction
Classifies each finding as Reflects or Contradicts the corpus
Computes overall and group-level contradiction rates
Generates the RQ3 alignment heatmap

#### Output files:

RQ3_corpus_differences.csv — corpus frequency rates
RQ3_comparison_results.csv — full alignment classifications
RQ3_full_comparison.csv — full comparison with race column

Estimated runtime: 30–60 minutes (no GPU required)

#### Troubleshooting
NameError: filtered_agency is not defined
Run Kernel → Restart & Run All to execute all cells in order.
NameError: mlm_sig_plot is not defined
Same fix — run Kernel → Restart & Run All.
RQ3 cannot find result files
Ensure Steps 1 and 2 completed successfully and update file paths in RQ3 Cell 6.
ClinicalBERT download fails
Check internet connection. Model is approximately 440MB.
Memory error
Reduce max_per_word from 20 to 10 in the template extraction function.
